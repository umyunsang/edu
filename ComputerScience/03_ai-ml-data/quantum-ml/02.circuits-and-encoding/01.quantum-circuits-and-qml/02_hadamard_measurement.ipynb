# Day11 Hadamard 중첩 측정 실습
구글 드라이브 공유 실습 소스를 Colab 실행용 노트북으로 변환한 자료입니다.


```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1)

qc.h(0)

qc.measure_all()

sim = AerSimulator()

job = sim.run(qc, shots=1000)

result = job.result()

counts = result.get_counts()

print(counts)
```
