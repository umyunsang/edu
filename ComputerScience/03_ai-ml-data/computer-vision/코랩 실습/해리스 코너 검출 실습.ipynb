# 해리스 코너 검출 실습 (Harris Corner Detection)

이 실습은 이미지 내에서 특징점(Feature Point) 중 하나인 **코너(Corner)**를 검출하는 대표적인 알고리즘인 **해리스 코너 검출** 과정을 다룹니다. 영상의 변화량이 모든 방향으로 큰 지점을 코너로 식별합니다.

```python
import cv2
import numpy as np
import urllib.request
from google.colab.patches import cv2_imshow

# 1. 샘플 이미지 다운로드 (스도쿠 이미지)
url = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/data/sudoku.jpg'
urllib.request.urlretrieve(url, 'sudoku.jpg')
img = cv2.imread('sudoku.jpg')

# 2. 코너 검출 전처리
# 코너 검출은 그레이스케일 이미지에서 수행함
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# 계산 정밀도를 위해 float32 타입으로 변환
gray = np.float32(gray)

# 3. 해리스 코너 검출 수행
# cv2.cornerHarris(입력영상, 블록크기, 소벨커널크기, k값)
# - 2: 블록 크기 (Block Size). 코너 검출을 위한 윈도우 크기
# - 3: 소벨 커널 크기 (Sobel Kernel Size). 미분 값을 계산할 때 사용
# - 0.04: Harris 코너 검출 방정식의 k값 (보통 0.04~0.06 사용)
dst = cv2.cornerHarris(gray, 2, 3, 0.04)

# 4. 결과 보정
# 검출된 코너 점들을 시각적으로 뚜렷하게 만들기 위해 팽창(Dilation) 수행
dst = cv2.dilate(dst, None)

# 5. 임계값(Threshold) 적용 및 시각화
# 영상 내 가장 강한 코너 점의 값(dst.max())의 1%보다 큰 픽셀만 코너로 간주
img_result = img.copy()
threshold = 0.01 * dst.max()
img_result[dst > threshold] = [0, 0, 255]  # 임계값을 넘는 좌표를 빨간색으로 표시

# 6. 결과 출력
print("--- [원본 이미지] ---")
cv2_imshow(img)
print("\n--- [해리스 코너 검출 결과 (빨간색 점)] ---")
cv2_imshow(img_result)
```
