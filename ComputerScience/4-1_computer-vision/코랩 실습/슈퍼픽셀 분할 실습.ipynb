# 슈퍼픽셀 분할 실습 (SLIC)

이 실습은 **SLIC(Simple Linear Iterative Clustering)** 알고리즘을 사용하여 이미지를 의미 있는 최소 단위인 **슈퍼픽셀(Superpixel)**로 분할하는 과정을 다룹니다. 슈퍼픽셀은 연산 효율을 높이고 이미지의 구조적 정보를 보존하는 데 유용합니다.

```python
import cv2
import numpy as np
import urllib.request
import matplotlib.pyplot as plt
from skimage.segmentation import slic, mark_boundaries
from skimage.color import label2rgb

# 1. 샘플 이미지 다운로드
url = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg'
urllib.request.urlretrieve(url, 'butterfly.jpg')
img = cv2.imread('butterfly.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 2. SLIC 알고리즘 적용
# n_segments: 분할할 슈퍼픽셀의 개수
# compactness: 색상 유사도와 공간적 거리의 가중치 (높을수록 사각형에 가까워짐)
segments = slic(img_rgb, n_segments=400, compactness=10, sigma=1, start_label=1)

# 3. 각 슈퍼픽셀 영역을 해당 영역의 평균 색상으로 채우기
superpixel_avg = label2rgb(segments, img_rgb, kind='avg')

# 4. 결과 시각화
plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.title("1. Original (Butterfly)")
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("2. SLIC Boundaries (400 segments)")
plt.imshow(mark_boundaries(img_rgb, segments))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("3. Segmented (Avg Color)")
plt.imshow(superpixel_avg.astype('uint8'))
plt.axis('off')

plt.tight_layout()
plt.show()
```
